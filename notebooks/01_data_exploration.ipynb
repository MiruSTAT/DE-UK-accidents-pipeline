### Checking for duplicates

```python
%sql
-- Checking Vehicle table using a composite key Accident_Index + Vehicle_Reference
SELECT Accident_Index, Vehicle_Reference, COUNT(*) as cnt
FROM workspace.default.vehicles_0514
GROUP BY Accident_Index, Vehicle_Reference
HAVING COUNT(*) > 1;
```

```python
%sql
--Checking Accidents table using Accident_Index
SELECT 
Accident_Index, 
count(*)
FROM accidents_0514
GROUP BY Accident_Index
HAVING count(*) > 1
```

```python
%sql
--Checking Casualties using coposite key = Accident_Index + Casualties_Reference + Vehicle_Reference
SELECT
Accident_Index,
Casualty_Reference,
Vehicle_Reference,
count(*)
FROM workspace.default.Casualties_0514
GROUP BY Accident_Index, Casualty_Reference, Vehicle_Reference
HAVING count(*) > 1;
```

```python
%sql
SELECT v.Accident_Index
FROM workspace.default.vehicles_0514 v
LEFT JOIN workspace.default.accidents_0514 a 
  ON v.Accident_Index = a.Accident_Index
WHERE a.Accident_Index IS NULL;
```

```python
%sql
SELECT c.Accident_Index
FROM workspace.default.casualties_0514 c
LEFT JOIN workspace.default.accidents_0514 a 
  ON c.Accident_Index = a.Accident_Index
WHERE a.Accident_Index IS NULL;
```

```python
%sql
SELECT c.Accident_Index, c.Vehicle_Reference
FROM workspace.default.casualties_0514 c
LEFT JOIN workspace.default.vehicles_0514 v 
  ON c.Accident_Index = v.Accident_Index 
  AND c.Vehicle_Reference = v.Vehicle_Reference
WHERE v.Accident_Index IS NULL;
```
